***d) Change the data type of all numeric columns. Find a way to deal with the percentage-columns and comma.***

```python
# Changing data types

repl_comma = ['Population', 'Net_Change', 'Density', 'Area', 'Net_Migrants', 'Median_Age']
repl_perc = ['Yearly_%_Change', 'Urban_Pop_%', 'World_Share']
repl_none = ['Fert_Rate']


def change_data_type(df, col_list, to_repl):
    
    if to_repl == 'comma':
        for col in col_list:
            df[col] = df[col].str.replace(",", "").astype(int) 
            
    elif to_repl == 'percent':
        for col in col_list:
            df[col] = df[col].str.replace("%", "").astype(float) 
    
    elif to_repl == 'none':
        for col in col_list:
            df[col] = df[col].astype(float)
            
    else:
        print("Please specify, what should be replaced")
        
    return df
```

***b) List Comprehension*** 

We’ve also learned about list comprehension. Here, the syntax was:
`L = [<element> for <element> in <list> if <Condition>]`

Use list comprehensions to write a Python function `remove_long_words()` that:
-	accepts a sentence and an integer n as input parameters
-	uses the split() function of String objects to split the sentence into words
-	stores the individual words in a list
-	removes all words that are longer than n characters from the list, thereby creating a new list
-	prints the new list


**2 Punkte**

```python
# Your solution can go here:

def remove_long_words(sentence, length): 
    """
    Function for dropping too long words from a sentence
    
    Parameters:
    -------------
    s: sentence to be split into words (string format)
    n: max. number of characters for a word in the sentence (integer format)
    
    Returns:
    -------------
    result: list, storing the words of the input sentence s 
    that are shorter (or equal) in characters than n
    
    """
    
    # Splitting the sentence into a list of words
    word_list = sentence.split()
    
    # Apply list-comprehension, filterin for words <= length
    
    return [word for word in word_list if len(word) <= length]


# Test case
remove_long_words("Hello I am Elena", 3)
```

```python
# Lösung von einem Studi

def count_bits(n):
    
    """
    Function counts non-zero bits 
    
    Parameters
    ----------
    n: Integer
        Non-negative number
    
    Returns
    -------
    Integer  
        Number of bins  
    
    """
    
    if n >=0: # the function should only work for non-negative values
        x = 0 # create x for the number of bins, with zero counted bins at the beginning 
        while n > 0: # do this loop as long it is larger than zero 
            x = x + (n & 1) # checks if the most right bit is one or zero and if it is one it adds 1 to the count x
            n = n >> 1 # moves n by one bit 
        return x

    else:
        print('This number is not convertable.') # if the number is negative this shows a nice message instead of an error. 
    

# test
print(count_bits(13))
```
